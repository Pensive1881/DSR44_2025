# Refactoring Example - Step 2: Packing in a class

```mermaid
flowchart TD
    A[Start Pipeline] --> B[Load Data]
    B --> C[Validate Schema]

    C --> D[Clean Data]
    
    D --> F[Transform Data]
    
    F --> G[Write to Database]

    G --> H[End Pipeline]
```

````python
import pandas as pd
import sqlite3

class DataCleaningPipeline:
    def run(self, filepath: str):
        # Load data
        df = pd.read_csv(filepath)
        print("Data loaded successfully.")

        # Validate columns
        if not all(col in df.columns for col in ["id", "name", "email", "signup_date"]):
            print("Missing some columns...")

        # Clean data - Data Cleaning
        df["email"] = df["email"].str.lower()
        df["signup_date"] = pd.to_datetime(df["signup_date"])
        df = df.dropna(subset=["id", "email"])
        print("Cleaning done.")

        # Transform data - Feature Engineering
        df["days_since_signup"] = pd.Timestamp.now() - df["signup_date"].dt.days

        # Save to database
        conn = sqlite3.connect("data/cleaned_customers.db")
        df.to_sql("customers", conn, if_exists="replace", index=False)
        conn.close()
        print("Data written to database.")

        print("Pipeline finished.")



filepath="data/customers.csv"
````
