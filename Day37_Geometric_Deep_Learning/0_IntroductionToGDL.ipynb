### Groups

Group theory has been developed hand in hand with geometries and it plays a crucial role in the studies of symmetries.

**Group theory is a branch of mathematics that studies the algebraic structures known as groups**

- A group is a set $G$ combined with a binary operation $\cdot$ (reffered as multiplication) that satisfies the following four axioms:
    1. **Closure:** For every pair of elements $a$, $b \in G$, the result of the operation $a \cdot b$ is also an element of $G$, i.e. $a \cdot b \in G$.
    2. **Associativity:** The group operation is associative, meaning that for all $a, b, c \in G$ the following holds: $(a \cdot b) \cdot c = a \cdot (b \cdot c)$.
    3. **Identity element:** There exists an element $e \in G$, called the identity, such that for every element $a \in G$ the equation $a \cdot e = a$ holds.
    4. **Inverse element:** For every element $a \in G$, there exists an element $b \in G$, called the inverse, such that the equation $a \cdot b = b \cdot a = e$, where $e$ is the identity element, holds.


#### SO(3) group

- [SO(3) group](https://en.wikipedia.org/wiki/3D_rotation_group) is a Special Orthogonal group of all proper (not including reflections) rotations around the origin of three-dimensional Euclidean space $R^3$.
- Proper rotation means that the determinant $det(g) = 1$, where $g$ is an element of the group SO(3), i.e. an orthogonal rotation matrix.
- An example element of this group is a rotation matrix around z-axis.

$$
R_z(\theta) = \begin{pmatrix}
\cos \theta & -\sin \theta & 0 \\
\sin \theta & \cos \theta & 0 \\
0 & 0 & 1
\end{pmatrix}
$$

This matrix represents a rotation of a vector in $\mathbb{R}^3$ by an angle $\theta$ about the z-axis. For example, a rotation of 90 degrees ( $\theta = \pi/2 $) would be:

$$
R_z\left(\frac{\pi}{2}\right) = \begin{pmatrix}
0 & -1 & 0 \\
1 & 0 & 0 \\
0 & 0 & 1
\end{pmatrix}
$$


#### Permutation group

- A [permutation group](https://en.wikipedia.org/wiki/Permutation_group) is a group whose elements are permutations of a given set.
- Consider a set of elements {1, 2, 3}. One possible permutation can be {2, 3, 1} and it is given by the multiplication of the initial vector $(1, 2, 3)$ representing the set with the following permutation matrix:

$$
P = \begin{pmatrix}
0 & 1 & 0 \\
0 & 0 & 1 \\
1 & 0 & 0
\end{pmatrix},
$$

where the permutation matrix $P$ is an element of the permuation group for this set.

# Symmetries and Geometric Priors in Machine Learning

## The curse of dimensionality

**The curse of dimensionality means that the amount of data needed to accurately fit a function grows exponentially with the number of dimensions**

- If you need 10 points to cover a 1D interval adequately, you might need 100 points for a 2D area, 1,000 points for a 3D volume, and so on.
- Learning in many dimensions is intractable in general.

<img src = "https://drive.google.com/uc?id=1AqWj7pCcsSK37CHKW9G9US6jAxGC1nVD" width="700" height="400" align="center">

The deep learning answer to the curse of dimensionality is the usage of geometric architectures that preserve particular symmetries, making it possible to learn the underlying multidimensional data distribution with some function like a neural network.

The learning becomes tractable when one makes use of 2 principles:

- **Symmetries.**
- **Scale separation.**

### **Using these 2 techniques one can substantially narrow the class of possible functions that can efficiently represent and describe a given signal on a domain.**

### Invariance VS equivariance:

- Consider a problem of a cat classifier neural network that predicts if there's a cat in the image.
- The data lies on a 2D-grid domain in this case.
- A 2D-grid exhibits translational symmetry, since the grid structure stays unchanged if the grid is shifted.

Therefore, we can exploit this symmetry implicitly by making sure that the model respects invariance and equivariance to translations.

<img src = "https://drive.google.com/uc?id=1EIDxH8DrbS3fsJ7HU5sZ8LFbRJKO8SB_" width="700" height="400" align="center">

Why do we need both?

- To recognize whole objects, we need to first recognize parts. This is why neural networks should be deep, hence we do hierarchical feature learning.
- If we make the intermediate representations invariant, we lose critical information.

## Scale separation

Creating architectures that make use of the inherent symmetries on a given domain is not enough to overcome the curse of dimensionality problem. The second main component is the **scale separation** technique.

Modern deep learning architectures are *composite*, meaning they are composed of many similar layers that learn hierarchy of features, just as humans use *compositionality* to give representation and meaning to complex ideas. That leads to the assuption that:

**Compositionality is useful for describing the world around us!**

### Scale separation as formalization of compositionality

The idea of scale separation is used in many areas of science, especially in physics.

<img src = "https://drive.google.com/uc?id=1bWOrnzmnEh5IxKogWVG9Aa-qJBkk6Myt" align="center">

Scale separation helps with understanding different phenomena, breaking them down into problems that can be addressed at different scales. For example, one can explain how planes fly using classical mechanics, but in order to explain subatomic particle interactions, one needs to use quantum mechanics that works with significantly smaller scales in space and time.

Similarly, in deep learning we usually reduce the scale of the problem by applying pooling layers inside the neural network.

**Making the resolution more coarse helps with the curse of dimensionality.**
