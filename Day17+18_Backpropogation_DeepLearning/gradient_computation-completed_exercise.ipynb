```python
import torch

# We are to predict the true price of an electronic gadget (a laptop)


# target, aka ground truth value, real price for our use case
y = torch.tensor(1220.0, requires_grad =False)
y
```

```python
# Feature, weight in grams of the laptop
x = torch.tensor(16.0, requires_grad=False)
```

```python
# Initial value of the weight in the model (assume it was randomly chosen)
w = torch.tensor(0.711, requires_grad=True)

# Bias term
b = torch.tensor(0.11, requires_grad=True)
```

```python
# Inside PyTorch this is what we do inside nn.linear layers
y_hat = w * x + b
y_hat
```

```python
loss_value = torch.abs(y_hat - y)
loss_value
```
