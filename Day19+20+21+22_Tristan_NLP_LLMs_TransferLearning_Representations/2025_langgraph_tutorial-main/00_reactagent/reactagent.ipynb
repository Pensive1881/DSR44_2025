## Importing environment variables

Before running the next cell, make sure you have a `.env` file in the same directory as this notebook with the following variables:

```bash
# Antropic.
LANGCHAIN_CHAT_MODEL_ANTHROPIC=...
ANTHROPIC_API_KEY=...
```

We will start with Anthropic, but you can use any LLM provider supported by LangChain. Get your API key here: https://platform.claude.com/dashboard

Now we can run the next cell to load the environment variables.

````python
import os
import dotenv
dotenv.load_dotenv()
print("Environment variables loaded.")
print(os.environ["LANGCHAIN_CHAT_MODEL_ANTHROPIC"])
````

## Our first agent

We boldly go and start with the ReAct agent. We will later learn how it works under the hood.

````python
from langchain.chat_models import init_chat_model
from langgraph.prebuilt import create_react_agent

# Initialize the chat model.
model = init_chat_model(
    os.environ["LANGCHAIN_CHAT_MODEL_ANTHROPIC"],
)

# Create the react agent without any tools.
agent = create_react_agent(
    model=model,
    tools=[],
)

# Invoke the agent with a user message.
result = agent.invoke(
    { "messages": [
        {
            "role": "user",
            "content": "What are agentic workflows and what are agents according to Anthropic?",
        }
    ]}
)
print(result["messages"][-1].content)
````
